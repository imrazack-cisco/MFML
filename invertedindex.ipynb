```python
import re
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import nltk

# Download necessary resources (if not already downloaded)
# It's generally safer to call download directly for required resources.
# NLTK's download function is idempotent, meaning it won't re-download if already present.
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True) # Added to fix LookupError

# --- Step 1: Define Documents (3 sentences) ---
docs = {
    1: "Information Retrieval is the process of obtaining relevant information from large repositories.",
    2: "The retrieval process involves searching, indexing, and ranking of documents.",
    3: "Efficient information retrieval improves the performance of search engines."
}

print("Original Documents:")
for doc_id, text in docs.items():
    print(f"Doc {doc_id}: {text}")
print("-" * 30)

# --- Step 2: Preprocessing Function ---
def preprocess(text):
    # Remove special characters and keep only letters, numbers, and spaces
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

    # Tokenize and convert to lower case
    tokens = word_tokenize(text.lower())

    # Remove stop words
    stop_words = set(stopwords.words('english'))
    filtered = [w for w in tokens if w not in stop_words]

    # Stemming
    ps = PorterStemmer()
    stemmed = [ps.stem(w) for w in filtered]

    return stemmed

# --- Step 3: Apply Preprocessing ---
preprocessed_docs = {}
print("\nPreprocessed Documents:")
for doc_id, text in docs.items():
    terms = preprocess(text)
    preprocessed_docs[doc_id] = terms
    print(f"Doc {doc_id}: {terms}")
print("-" * 30)

# --- Step 4: Build Inverted Index ---
inverted_index = {}

for doc_id, terms in preprocessed_docs.items():
    for term in terms:
        if term not in inverted_index:
            # Initialize with a set to automatically handle duplicates within a document's terms
            inverted_index[term] = set()
        inverted_index[term].add(doc_id)

# Convert sets to sorted lists for final display/structure
for term in inverted_index:
    inverted_index[term] = sorted(list(inverted_index[term]))

# --- Step 5: Display Inverted Index ---
print("\nInverted Index (term -> document IDs):")
# Sort by term for clean display
for term in sorted(inverted_index.keys()):
    print(f"{term:<15} -> {inverted_index[term]}")
```
